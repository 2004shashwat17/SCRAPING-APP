```python
import pandas as pd
import numpy as np

path = "/content/shashwat_facebook_integrated_output.csv"
df = pd.read_csv(path)
```

```python
interaction_cols = [
    "who_commented",
    "what_comment",
    "who_liked",
    "who_shared",
    "shared_url"
]

numeric_cols = [
    "like_count",
    "comment_count",
    "share_count",
    "level"
]

# everything else is metadata
meta_cols = [
    col for col in df.columns
    if col not in interaction_cols + numeric_cols
]
```

```python
agg_dict = {}

# metadata → first value
for col in meta_cols:
    agg_dict[col] = "first"

# numeric → max (safe for post-level counts)
for col in numeric_cols:
    if col in df.columns:
        agg_dict[col] = "max"

# interactions → unique comma-separated
for col in interaction_cols:
    if col in df.columns:
        agg_dict[col] = lambda x: ",".join(
            sorted(set(x.dropna().astype(str)))
        )
```

```python
cleaned_post_df = (
    df
    .groupby("post_no", as_index=False)
    .agg(agg_dict)
)
```

```python
for col in ["like_count", "comment_count", "share_count"]:
    if col not in cleaned_post_df.columns:
        cleaned_post_df[col] = 0
```

```python
cleaned_post_df["post_score"] = (
    cleaned_post_df["like_count"] * 1 +
    cleaned_post_df["comment_count"] * 7 +
    cleaned_post_df["share_count"] * 14
)
```

```python
cleaned_post_df[
    ["post_no", "like_count", "comment_count", "share_count", "post_score"]
].head()
```

```python
final_path = "/content/cleaned_facebook_posts_with_post_score.csv"

cleaned_post_df.to_csv(
    final_path,
    index=False
)
```
